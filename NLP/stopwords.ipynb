```python
from nltk.stem import PorterStemmer,SnowballStemmer,WordNetLemmatizer
```

```python
from nltk.corpus import stopwords
```

```python
nltk.download('stopwords')
```

```python
kalam_speech = (
    "Dreams are not that which you see while sleeping, but dreams are those which do not let you sleep. "
    "To succeed in your mission, you must have single-minded devotion to your goal. "
    "If you fail, never give up because FAIL means First Attempt In Learning. "
    "End is not the end; in fact, END means Effort Never Dies. "
    "If you get NO as an answer, remember NO means Next Opportunity. "
    "Let us sacrifice our today so that our children can have a better tomorrow. "
    "Have the courage to think differently, courage to invent, to travel the unexplored path, "
    "courage to discover the impossible, and to conquer the problems and succeed."
)
```

```python
snowballStemmer = SnowballStemmer(language = 'english')
```

```python
sentences = nltk.sent_tokenize(kalam_speech)
```

```python
for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [
        snowballStemmer.stem(word)
        for word in words
        if word.lower() not in set(stopwords.words("english"))
    ]
    sentences[i] = " ".join(words)
```
