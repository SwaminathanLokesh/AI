```python
import nltk
```

```python
paragraph = (
    "Natural language processing helps computers understand human language. "
    "Machine learning models analyze words, phrases, and grammar patterns. "
    "Proper text preprocessing improves the performance of every language model."
)
```

```python
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt_tab')
```

```python
words = nltk.sent_tokenize(paragraph)
```

```python
import sklearn
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
objOHE = OneHotEncoder()
```

```python
words_2d = [[word] for word in words]
one_hot_encoded_words = objOHE.fit_transform(words_2d)
print("Original words:", words)
print("One-Hot Encoded (sparse matrix):")
display(one_hot_encoded_words)
```

```python
one_hot_encoded_words
```

```python
dense_one_hot_encoded_words = one_hot_encoded_words.toarray()
print("Dense One-Hot Encoded Array:")
display(dense_one_hot_encoded_words)
```
