```python
test_words = [
    # Standard Suffixes
    "connect",
    "connected",
    "connecting",
    "connections",
    # Silent 'e' vs. Doubled Consonants
    "hop",
    "hopped",
    "hopping",
    "hope",
    "hoped",
    "hoping",
    # 'y' to 'i' Mutations
    "happy",
    "happily",
    "happiness",
    # Derivational Suffixes
    "activate",
    "active",
    "actively",
    "activity",
    "activation",
    # Degree & Comparison
    "fast",
    "faster",
    "fastest",
    # Irregular Forms
    "run",
    "ran",
    "running",
    "go",
    "went",
    "gone",
    # Over-Stemming Traps
    "university",
    "universal",
    "universe",
    # Under-Stemming Traps
    "alumnus",
    "alumni",
    "alumnae",
    # Additional common edge cases
    "generalization",
    "generalize",
    "organized",
    "organizing",
    "organization",
    "flies",
    "flying",
]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming = PorterStemmer()
```

```python
for word in test_words:
  print(word + "---->" + stemming.stem(word))
```

```python
from nltk.stem import RegexpStemmer
```

```python
custom_pattern = r"(ing|ed|ly|es|s)$"
regexp_stemmer = RegexpStemmer(custom_pattern, min=4)
```

```python
for word in test_words:
  print(word + "---->" + regexp_stemmer.stem(word))
```

```python
from nltk.stem import SnowballStemmer
```

```python
snowball_stemmer = SnowballStemmer(language='english')
```

```python
for word in test_words:
  print(word + "---->" + snowball_stemmer.stem(word))
```

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatization = WordNetLemmatizer()
```

```python
for word in test_words:
  print(word + "---->" + lemmatization.lemmatize(word,pos='v'))
```
