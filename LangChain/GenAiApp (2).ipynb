```python
from langchain_community.document_loaders import WebBaseLoader
```

```python
loaders = WebBaseLoader("https://docs.langchain.com/oss/python/deepagents/overview")
```

```python
loaders
```

```python
docs = loaders.load()
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size = 1000,
    chunk_overlap = 200
)
```

```python
chunks = text_splitter.split_documents(docs)
```

```python
import os
from langchain_chroma import Chroma
from langchain_google_genai import GoogleGenerativeAIEmbeddings

# 1. Initialize embeddings with the active model
embeddings = GoogleGenerativeAIEmbeddings(
    model="models/gemini-embedding-001"
)

# 2. Add chunks to ChromaDB
vector_store = Chroma.from_documents(
    documents=chunks,            # your split chunks list
    embedding=embeddings,
    collection_name="website_docs"
)

print("Chunks embedded and stored successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_google_genai import ChatGoogleGenerativeAI
# Import from langchain_classic instead of langchain
from langchain_classic.chains.combine_documents import create_stuff_documents_chain
from langchain_classic.chains import create_retrieval_chain

# 1. Initialize Gemini LLM
llm = ChatGoogleGenerativeAI(model="gemini-3.8-flash")

# 2. Define the Chat Prompt Template
prompt = ChatPromptTemplate.from_template("""
Answer the following question based only on the provided context:

<context>
{context}
</context>

Question: {input}
""")

# 3. Create document chain
document_chain = create_stuff_documents_chain(llm, prompt)
print("Prompt template and document chain created!")
```
