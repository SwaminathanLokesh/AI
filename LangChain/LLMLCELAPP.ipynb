```python
import os
from google.colab import userdata
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain_core.messages import SystemMessage, HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
os.environ["GOOGLE_API_KEY"] = userdata.get('GEMINI_API_KEY')
model = ChatGoogleGenerativeAI(
    model="gemini-3.5-flash-lite",
    temperature=0.3,
    max_retries=3  # Automatically handles brief burst limits with exponential backoff
)
```

```python

parser = StrOutputParser()
chain = model | parser
```

```python
language = input("Enter target language: ")
text = input("Enter text to translate: ")
```

```python
messages = [
    SystemMessage(content=f"You are a professional translator. Translate the given text accurately to {language}."),
    HumanMessage(content=text)
]
```

```python
result = chain.invoke(messages)
```

```python
from langchain_core.prompts import ChatPromptTemplate
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a professional translator. Translate the given text accurately to {language}."),
    ("user", "{text}")
])
```

```python
chain = prompt | model | parser
```

```python
user_lang = input("Enter target language: ")
user_text = input("Enter text to translate: ")
```

```python
result = chain.invoke({
    "language": user_lang,
    "text": user_text
})
```
