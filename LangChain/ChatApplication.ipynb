```python
from langchain_community.chat_message_histories import ChatMessageHistory
from langchain_core.chat_history import BaseChatMessageHistory
from langchain_core.runnables.history import RunnableWithMessageHistory
```

```python
store = {}

def get_session_history(session_id: str) -> BaseChatMessageHistory:
    if session_id not in store:
        store[session_id] = ChatMessageHistory()
    return store[session_id]
```

```python
with_message_history = RunnableWithMessageHistory(
    llm,
    get_session_history
)
```

```python
config = {"configurable": {"session_id": "chat1"}}
```

```python
response_1 = with_message_history.invoke(
    [HumanMessage(content="Hey, my favorite food is biryani.")],
    config=config
)
print("Response 1:", response_1.content)
```

```python
response_2 = with_message_history.invoke(
    [HumanMessage(content="What is my favorite food?")],
    config=config
)
```

```python
print("\nResponse 2:", response_2.content)
```

```python
config_chat1 = {"configurable": {"session_id": "chat1"}}
config_chat2 = {"configurable": {"session_id": "chat2"}}
```

```python
from langchain_core.prompts import ChatPromptTemplate, MessagesPlaceholder
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful and witty AI assistant. Answer queries concisely and clearly."),
    MessagesPlaceholder(variable_name="messages"),
])
```

```python
chain = prompt | llm
```

```python
response = chain.invoke({
    "messages": [HumanMessage(content="Explain what an API is in one short sentence.")]
})
```

```python
print(response.content)
```

```python
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)
```

```python
from operator import itemgetter
from langchain_core.messages import trim_messages
```

```python
trimmer = trim_messages(
    max_tokens=45,                  # Adjust token limit as needed
    strategy="last",                # Keep the most recent messages
    token_counter=llm,              # Uses the Gemini model instance to count tokens
    include_system=True,            # Keep system instructions intact
    allow_partial=False,            # Avoid cutting a single message in half
    start_on="human"                # Ensure trimmed history starts on a user message
)
```

```python
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage

messages = [
    SystemMessage(content="You are a helpful coding assistant."),
    HumanMessage(content="Hi! I am building a chat application."),
    AIMessage(content="That sounds exciting! What tech stack are you using?"),
    HumanMessage(content="I am using LangChain and Google Gemini."),
    AIMessage(content="Great choice! LangChain pairs really well with Gemini."),
    HumanMessage(content="What was the second technology I mentioned?"),
]
```

```python
from langchain_core.runnables import RunnablePassthrough
```

```python
# Chain that takes incoming input dict, trims the messages via itemgetter,
# preserves remaining keys (like language/extra params), formats via prompt, and invokes llm
chain = (
    RunnablePassthrough.assign(
        messages=itemgetter("messages") | trimmer
    )
    | prompt
    | llm
)
```

```python
response = chain.invoke({
    "messages": messages + [HumanMessage(content="Can you summarize what we discussed so far?")]
})

print(response.content)
```

```python
# 1. Wrap the trimming chain with session-based message history
with_message_history = RunnableWithMessageHistory(
    chain,
    get_session_history,
    input_messages_key="messages"
)

# 2. Define session config
config = {"configurable": {"session_id": "chat_trimmed_session"}}
```
