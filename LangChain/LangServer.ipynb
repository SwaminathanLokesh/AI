```python
import os
import warnings
warnings.filterwarnings("ignore")
```

```python
from fastapi import FastAPI
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
from langserve import add_routes
```

```python
from google.colab import userdata
os.environ["GEMINI_API_KEY"] = userdata.get("GEMINI_API_KEY")
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite", temperature=0.3)
```

```python
system_template = "You are a helpful assistant that translates {input_language} to {output_language}."
prompt_template = ChatPromptTemplate.from_messages([
    ("system", system_template),
    ("user", "{text}")
])
```

```python
parser = StrOutputParser()
```

```python
chain = prompt_template | model | parser
```

```python
app = FastAPI(
    title="LangChain Server",
    version="1.0",
    description="API server using LangChain, LangServe, and Gemini Lite"
)
```

```python
add_routes(
    app,
    chain,
    path="/chain"
)
```

```python
import threading
import uvicorn
```

```python
import threading
import uvicorn
from google.colab import output

# Start Uvicorn cleanly on port 8000
def run_app():
    uvicorn.run(app, host="127.0.0.1", port=8000, log_level="warning")

server_thread = threading.Thread(target=run_app, daemon=True)
server_thread.start()

print("Server active! Rendering LangServe playground below:")

# Render the LangServe playground interface inside the notebook cell
output.serve_kernel_port_as_iframe(8000, path="/chain/playground/")
```
