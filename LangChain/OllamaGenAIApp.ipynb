```python
import streamlit as st
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful and knowledgeable AI assistant. Please provide concise, accurate answers to the user's questions."),
    ("user", "Question: {question}")
])
```

```python
st.title("Local GenAI Assistant with Ollama")
input_text = st.text_input("What is your question?")
```

```python

import subprocess
import time

# 1. Start Ollama in the background
subprocess.Popen(["ollama", "serve"])

# 2. Give the daemon a few seconds to initialize
time.sleep(5)

# 3. Pull your lightweight model
!ollama pull gemma:2b
```

```python
from langchain_ollama import ChatOllama
llm = ChatOllama(
    model="gemma:2b",  # or "llama3", "llama2", etc.
    temperature=0.3
)

# 3. Output Parser
output_parser = StrOutputParser()

# 4. Chain using LCEL
chain = prompt | llm | output_parser

# 5. Trigger on input
if input_text:
    with st.spinner("Generating response..."):
        response = chain.invoke({"question": input_text})
        st.write(response)
```
